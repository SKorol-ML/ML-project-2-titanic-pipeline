```python
import sys
import os
sys.path.append(os.path.abspath('..'))

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# Здесь позже будем импортировать наши функции
print("Готово")
```

```python
def load_data(filepath):
    """
    Загружает данные из CSV-файла и возвращает DataFrame.
    Если файл не найден, выбрасывает исключение.
    """
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"Файл {filepath} не найден")
    return pd.read_csv(filepath)

# Проверяем
df = load_data('../data/train.csv')
print(df.shape)
df.head()
```

```python
def split_data(X, y, test_size=0.2, random_state=42, stratify=True):
    """
    Разделяет данные на обучающую и тестовую выборки.
    Если stratify=True, сохраняет пропорции классов (по умолчанию).
    Возвращает X_train, X_test, y_train, y_test.
    """
    if stratify:
        strat = y
    else:
        strat = None
    return train_test_split(X, y, test_size=test_size, random_state=random_state, stratify=strat)
```

```python
X = df.drop('Survived', axis=1)
y = df['Survived']

X_train, X_test, y_train, y_test = split_data(X, y)
print(f"Train: {X_train.shape}, Test: {X_test.shape}")
print(f"Баланс классов в train: {y_train.value_counts(normalize=True)}")
print(f"Баланс классов в test:  {y_test.value_counts(normalize=True)}")
```

```python
def save_predictions(y_pred, filepath='../predictions.csv', ids=None):
    """
    Сохраняет предсказания в CSV-файл.
    Если ids не переданы, использует порядковый номер.
    """
    if ids is None:
        ids = np.arange(len(y_pred))
    df_out = pd.DataFrame({'PassengerId': ids, 'Survived': y_pred})
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    df_out.to_csv(filepath, index=False)
    print(f"Предсказания сохранены в {filepath}")
```

```python
# Пример
y_pred_dummy = y_test  # просто для теста
save_predictions(y_pred_dummy, '../predictions_test.csv')
```
